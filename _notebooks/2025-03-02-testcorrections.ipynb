## Reverse Order Traversal on Arrays

```java
// Reverse traversal of a standard array
int[] numbers = {1, 2, 3, 4, 5};
for (int i = numbers.length - 1; i >= 0; i--) {
    System.out.println(numbers[i]);
}

// Reverse traversal of ArrayList using index
ArrayList<String> names = new ArrayList<>(Arrays.asList("Alice", "Bob", "Charlie"));
for (int i = names.size() - 1; i >= 0; i--) {
    System.out.println(names.get(i));
}

// Using ListIterator for reverse traversal
ArrayList<String> colors = new ArrayList<>(Arrays.asList("Red", "Green", "Blue"));
ListIterator<String> iterator = colors.listIterator(colors.size());
while (iterator.hasPrevious()) {
    System.out.println(iterator.previous());
}
```

## Deleting from ArrayList in a Loop

```java
// WRONG way (causes issues with index)
ArrayList<Integer> numbers = new ArrayList<>(Arrays.asList(1, 2, 3, 4, 5, 6));
for (int i = 0; i < numbers.size(); i++) {
    if (numbers.get(i) % 2 == 0) {
        numbers.remove(i);  // This will skip elements!
    }
}

// CORRECT way 1: Traverse backwards
ArrayList<Integer> numbers = new ArrayList<>(Arrays.asList(1, 2, 3, 4, 5, 6));
for (int i = numbers.size() - 1; i >= 0; i--) {
    if (numbers.get(i) % 2 == 0) {
        numbers.remove(i);  // Safe when going backwards
    }
}

// CORRECT way 2: Using Iterator
ArrayList<Integer> numbers = new ArrayList<>(Arrays.asList(1, 2, 3, 4, 5, 6));
Iterator<Integer> iterator = numbers.iterator();
while (iterator.hasNext()) {
    if (iterator.next() % 2 == 0) {
        iterator.remove();  // Safe removal with iterator
    }
}

// CORRECT way 3: Using removeIf (Java 8+)
ArrayList<Integer> numbers = new ArrayList<>(Arrays.asList(1, 2, 3, 4, 5, 6));
numbers.removeIf(n -> n % 2 == 0);  // Remove all even numbers
```

## Array Length vs ArrayList Size

```java
// Array length - fixed after creation
int[] array = new int[5];
int length = array.length;  // 5
System.out.println("Array length: " + length);

// ArrayList size - dynamic
ArrayList<String> list = new ArrayList<>();
list.add("Item 1");
list.add("Item 2");
int size = list.size();  // 2
System.out.println("ArrayList size: " + size);
```

## Static Variables and Methods

```java
public class MathHelper {
    // Static variable - shared across all instances
    public static final double PI = 3.14159;
    public static int counter = 0;
    
    // Static method - can be called without creating an instance
    public static int square(int num) {
        return num * num;
    }
    
    // Static method using static variable
    public static void incrementCounter() {
        counter++;
    }
}

// How to use static elements
public class StaticDemo {
    public static void main(String[] args) {
        // Access static variable without creating object
        System.out.println("PI value: " + MathHelper.PI);
        
        // Call static method without creating object
        int result = MathHelper.square(5);
        System.out.println("5 squared: " + result);
        
        // Using static counter
        MathHelper.incrementCounter();
        System.out.println("Counter: " + MathHelper.counter);  // 1
        MathHelper.incrementCounter();
        System.out.println("Counter: " + MathHelper.counter);  // 2
    }
}
```
